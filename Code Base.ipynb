```python
#Promotion Logic 

#no bonus
bonus_discount = 0.0
#plus Member bonus 
if customer_tier.upper() == "PLUS" and basket_value >= 750.0:
    bonus_discount += 2.0
    
#high busket value
if basket_value >= 1500.0:
    bonus_discount += 2.0

# Total discount cannot exceed 15%
total_discount = promo_pct + bonus_discount
if total_discount > 15.0:
    total_discount = 15.0


while True:
        #calculate the order
    net_merchandise_value = basket_value * (1 - total_discount / 100)
    
    last_mile_cost = 12 + (6 * distance_km)
    
    payment_tech_cost = 0.01 * net_merchandise_value
    
    contribution = (
        net_merchandise_value
        - cogs
        - 18
        - last_mile_cost
        - payment_tech_cost
    )
    
    contribution_margin = (contribution / net_merchandise_value) * 100

    if contribution_margin >= 5.0 or bonus_discount <= 0:
        break
    # Reduce ONLY discretionary bonus
    bonus_discount -= 1.0

    # Recalculate total discount
    total_discount = promo_pct + bonus_discount


print("Requested discretionary bonus:", bonus_discount, "%")
print("Final total discount:", total_discount, "%")
print("Net merchandise value:", round(net_merchandise_value, 2))
print("Last-mile cost:", round(last_mile_cost, 2))
print("Payment/technology cost:", round(payment_tech_cost, 2))
print("Contribution:", round(contribution, 2))
print("Contribution margin:", round(contribution_margin, 2), "%")
```

```python
raw_feed = """
QC-501|plus|1280.00|900.00|8|paid|ready|2.8|24|4;
QC-502|regular|760.00|560.00|5|paid|ready|4.9|18|9;
N/A;
QC-503|plus|1850.00|1350.00|10|paid|ready|3.2|31|6;
QC-504|regular|990.00|730.00|12|failed|ready|2.0|15|3;
QC-505|plus|1450.00|1120.00|9|paid|oops|1.7|20|2;
BAD_RECORD;
QC-506|regular|2100.00|1500.00|6|paid|ready|6.3|22|13;
QC-507|plus|890.00|760.00|12|paid|ready|2.1|82|5;
QC-509|regular|1700.00|1300.00|4|paid|ready|8.6|19|8;
QC-510|plus|1600.00|1300.00|11|paid|ready|4.2|20|7;
STORE_CLOSED;
QC-508|plus|2400.00|1600.00|8|paid|ready|1.2|12|2
"""

valid_orders = 0
skipped_records = 0
demand_gmv = 0.0
remaining_rider_slots = 4
dispatched_orders = 0
capacity_holds = 0
dispatched_net_value = 0.0
dispatched_contributon = 0.0
discision_counts = {}
commercial_conflict_order = 0
loss_making_orders = 0
highest_basket = 0.0
highest_order_id = ""

for index, value in enumerate(raw_feed.split(";")):
    value = value.strip()

    if value == "STORE_CLOSED":
        break
    if value in ["N/A", "BAD_RECORD"]:
        continue

    data = value.split("|")
    if len(data) != 10:
        skipped_records += 1
        continue

    order_id = data[0].strip()
    customer_tier = data[1].strip().upper()
    basket_value = float(data[2].strip())
    cogs = float(data[3].strip())
    promo_pct = float(data[4].strip())
    payment_status = data[5].strip().upper()
    stock_status = data[6].strip().upper()
    distance_km = float(data[7].strip())
    risk_score = int(data[8].strip())
    rider_wait_minutes = int(data[9].strip())

    valid_orders+=1
    demand_gmv+= basket_value

    bonus_discount = 0.0
    #plus Member bonus
    if customer_tier == "PLUS" and basket_value >= 750.0:
        bonus_discount += 2.0

    #high busket value
    if basket_value >= 1500.0:
        bonus_discount += 2.0

    # Total discount cannot exceed 15%
    total_discount = promo_pct + bonus_discount
    if total_discount > 15.0:
        total_discount = 15.0


    while True:
        #calculate the order
        net_merchandise_value = basket_value * (1 - total_discount / 100)

        delivery_cost = 12 + (6 * distance_km)

        payment_tech_cost = 0.01 * net_merchandise_value

        contribution = (
            net_merchandise_value
            - cogs
            - 18
            - delivery_cost
            - payment_tech_cost
        )

        contribution_margin = (contribution / net_merchandise_value) * 100

        if contribution_margin >= 5.0 or bonus_discount <= 0:
            break
        # Reduce ONLY discretionary bonus
        bonus_discount -= 1.0

        # Recalculate total discount
        total_discount = promo_pct + bonus_discount


    if contribution_margin >= 12.0:
        margin_signal = "HEALTHY"
    elif contribution_margin >= 5.0:
        margin_signal = "THIN"
    else:
        margin_signal = "LOSS MAKING"


    if payment_status.upper() != "PAID":
        fulfilment_status = "REJECT PAYMENT"
    elif stock_status.upper() != "READY":
        fulfilment_status = "REJECT STOCK"
    elif risk_score >= 75:
        fulfilment_status = "MANUAL_REVIEW"
    elif distance_km > 8:
        fulfilment_status = "OUTSIDE_SERVICE"
    else:
        fulfilment_status = "ELIGIBLE_FOR_RIDER"


    if fulfilment_status == "ELIGIBLE_FOR_RIDER":
        if remaining_rider_slots > 0:
            remaining_rider_slots -= 1
            if distance_km >6 or rider_wait_minutes >12:
                fulfilment_status = "DISPATCH_SLA_RISK"
            else:
                fulfilment_status = "READY_TO_DISPATCH"

                dispatched_orders += 1
                dispatched_net_value += net_merchandise_value
                dispatched_contributon += contribution
        else:
            fulfilment_status = "CAPACITY_HOLD"
            capacity_holds += 1



    discision_counts[fulfilment_status] = discision_counts.get(fulfilment_status, 0)+ 1

    if margin_signal== "HEALTHY" and fulfilment_status not in ["READY_TO_DISPATCH", "DISPATCH_SLA_RISK"]:
      commercial_conflict_order +=1
    if margin_signal== "LOSS_MAKING" and fulfilment_status in ["READY_TO_DISPATCH", "DISPATCH_SLA_RISK"]:
      loss_making_orders += 1

    if basket_value > highest_basket:
      highest_basket = basket_value
      highest_order_id = order_id

    print(f'''
    Position: {index+1}
    Order ID: {order_id}
    Final Dicount: {total_discount}%
    Margin: {contribution_margin:.2f}%
    Commercial Band: {margin_signal}
    Operational Decision: {fulfilment_status}
    Rider Slots Remaining: {remaining_rider_slots}
    ''')
```
